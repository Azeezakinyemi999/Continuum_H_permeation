```python
from ase.spacegroup import crystal
from ase.neighborlist import natural_cutoffs, NeighborList
from ase.io import read, write
import numpy as np

# ── Build Cr2O3 bulk unit cell ────────────────────────────
a = 4.96
c = 13.59

bulk_cr2o3 = crystal(
    symbols=['Cr', 'O'],
    basis=[(0, 0, 0.347), (0.306, 0, 0.25)],
    spacegroup=167,
    cellpar=[a, a, c, 90, 90, 120]
)

print(f"Unit cell: {bulk_cr2o3.get_chemical_formula()}, {len(bulk_cr2o3)} atoms")
print(f"Unit cell matrix:\n{bulk_cr2o3.get_cell()}")

# ── Repeat to get coincidence supercell size ──────────────
slab = bulk_cr2o3.repeat([2, 2, 1])

print(f"\nAfter 2×2×1 repeat:")
print(f"  Atoms:  {len(slab)}")
print(f"  Cell matrix:\n{slab.get_cell()}")

# ── Convert hexagonal to orthogonal (without Niggli) ─────
# Extract true orthogonal dimensions from hexagonal cell
cell = slab.get_cell()
# For hexagonal: Lx = a[0], Ly = |b|, Lz = c[2]
Lx = cell[0, 0]
Ly = np.linalg.norm(cell[1])  # Take full length of b vector
Lz = cell[2, 2]

print(f"\nOrthogonal dimensions:")
print(f"  Lx = {Lx:.3f} Å")
print(f"  Ly = {Ly:.3f} Å")
print(f"  Lz = {Lz:.3f} Å")

# Set orthogonal cell and scale positions
slab.set_cell(np.diag([Lx, Ly, Lz]), scale_atoms=True)
slab.set_pbc([True, True, True])
slab.wrap()

# ── Verify ────────────────────────────────────────────────
cell = slab.get_cell()
z_pos = slab.get_positions()[:, 2]
thickness = z_pos.max() - z_pos.min()

print(f"\n=== Cr2O3 Bulk Slab ===")
print(f"Atoms:          {len(slab)}")
print(f"Formula:        {slab.get_chemical_formula()}")
print(f"Cell (Å):       {cell[0,0]:.3f} x {cell[1,1]:.3f} x {cell[2,2]:.3f}")
print(f"Thickness:      {thickness:.2f} Å")
print(f"PBC:            {slab.get_pbc()}")
print(f"Cell matrix (should be diagonal):\n{slab.get_cell()}")

# ── Verify Cr-O bond lengths ──────────────────────────────
cutoffs = natural_cutoffs(slab, mult=1.2)
nl = NeighborList(cutoffs, self_interaction=False, bothways=True)
nl.update(slab)
symbols   = slab.get_chemical_symbols()
positions = slab.get_positions()

cr_o_bonds = []
for i, sym in enumerate(symbols):
    if sym == 'Cr':
        indices, offsets = nl.get_neighbors(i)
        for j, offset in zip(indices, offsets):
            if symbols[j] == 'O':
                d = np.linalg.norm(
                    positions[j] + offset @ slab.get_cell() - positions[i])
                cr_o_bonds.append(d)

if cr_o_bonds:
    print(f"\nCr-O bond lengths:")
    print(f"  Min: {min(cr_o_bonds):.3f} Å  (expected ~1.97 Å)")
    print(f"  Max: {max(cr_o_bonds):.3f} Å  (expected ~2.02 Å)")
else:
    print("\n⚠ No Cr-O bonds found — check structure")

# ── Write files ───────────────────────────────────────────
lammps_file = 'cr2o3_perfect.lammps'
write(lammps_file, slab,
      format='lammps-data',
      atom_style='atomic',
      specorder=['Cr', 'O'])

with open(lammps_file, 'a') as f:
    f.write('\nMasses\n\n')
    f.write('1 51.996  # Cr\n')
    f.write('2 15.999  # O\n')

write('cr2o3_perfect.xyz', slab)

print(f"\n✓ Files written: cr2o3_perfect.lammps + cr2o3_perfect.xyz")
```

### HASTELLOY_N


```python
from ase.build import bulk
from ase.io import write
import numpy as np

# ── Step 2: FCC Ni base ──────────────────────────────────
a_Ni = 3.524  # Angstrom
base = bulk('Ni', 'fcc', a=a_Ni, cubic=True)

print(f"Unit cell:")
print(f"  Atoms:   {len(base)}")
print(f"  Cell:    {base.get_cell().diagonal()}")

# 3x3x8 repeat
slab = base.repeat([3, 3, 8])
N = len(slab)
cell = slab.get_cell()

print(f"\nAfter 3x3x8 repeat:")
print(f"  Atoms:     {N}")
print(f"  Cell (Å):  {cell[0,0]:.2f} x {cell[1,1]:.2f} x {cell[2,2]:.2f}")
print(f"  Thickness: {cell[2,2]:.2f} Å")

# ── Step 3: Composition wt% → at% ────────────────────────
n_Mo = int(0.107 * N)
n_Cr = int(0.088 * N)
n_Fe = int(0.041 * N)
n_C  = int(0.0038 * N)
n_Al = int(0.0106 * N)
n_B  = int(0.0005 * N)

total_solute = n_Mo + n_Cr + n_Fe + n_C + n_Al + n_B
n_Ni = N - total_solute

print(f"\nComposition:")
print(f"  Ni: {n_Ni}  ({100*n_Ni/N:.1f} at%)")
print(f"  Mo: {n_Mo}  ({100*n_Mo/N:.1f} at%)")
print(f"  Cr: {n_Cr}  ({100*n_Cr/N:.1f} at%)")
print(f"  Fe: {n_Fe}  ({100*n_Fe/N:.1f} at%)")
print(f"  C:  {n_C}   ({100*n_C/N:.2f} at%)")
print(f"  Al: {n_Al}  ({100*n_Al/N:.2f} at%)")
print(f"  B:  {n_B}   ({100*n_B/N:.3f} at%)")
print(f"  Total solute: {total_solute}")

# ── Random substitution ───────────────────────────────────
np.random.seed(42)
symbols = np.array(slab.get_chemical_symbols())
indices = np.random.permutation(N)
print(symbols)
print(indices)

idx = 0
for element, count in [('Mo', n_Mo), ('Cr', n_Cr),
                        ('Fe', n_Fe), ('C',  n_C),
                        ('Al', n_Al), ('B',  n_B)]:
    symbols[indices[idx:idx+count]] = element
    idx += count

slab.set_chemical_symbols(symbols.tolist())

# ── Step 4: Add vacuum ────────────────────────────────────
slab.center(vacuum=5.0, axis=2)

# ── Step 5: Verify ────────────────────────────────────────
cell = slab.get_cell()
z_pos = slab.get_positions()[:, 2]
thickness = z_pos.max() - z_pos.min()
vacuum_actual = cell[2,2] - thickness

print(f"\nFinal slab:")
print(f"  Cell (Å):  {cell[0,0]:.2f} x {cell[1,1]:.2f} x {cell[2,2]:.2f}")
print(f"  Thickness: {thickness:.1f} Å")
print(f"  Vacuum:    {vacuum_actual:.1f} Å")
print(f"  Formula:   {slab.get_chemical_formula()}")

# ── Step 6: Write output files ────────────────────────────
lammps_file = 'hastelloy_n_perfect.lammps'
specorder = ['Ni', 'Mo', 'Cr', 'Fe', 'C', 'Al', 'B']
write(lammps_file, slab,
      format='lammps-data',
      atom_style='atomic',
      specorder=specorder)

# Append Masses
with open(lammps_file, 'a') as f:
    f.write('\nMasses\n\n')
    f.write('1  58.693  # Ni\n')
    f.write('2  95.960  # Mo\n')
    f.write('3  51.996  # Cr\n')
    f.write('4  55.845  # Fe\n')
    f.write('5  12.011  # C\n')
    f.write('6  26.982  # Al\n')
    f.write('7  10.811  # B\n')

write('hastelloy_n_perfect.xyz', slab)

print(f"\n=== Files Written ===")
print(f"  hastelloy_n_perfect.lammps")
print(f"  hastelloy_n_perfect.xyz")
print(f"\n✓ Ready for OVITO visualization")
```

## Plan for L1 Coincidence Supercell
- Step 1 — Find the best coincidence pair by testing all repeat combinations and finding minimum strain.
- Step 2 — Rebuild Cr₂O₃ at the coincidence repeat.
- Step 3 — Rebuild Hastelloy N at the coincidence repeat.
- Step 4 — Stack them with 2 Å interface gap.
- Step 5 — Verify and visualize.

```python
import numpy as np

# Natural unit cell dimensions after surface() call
# Cr2O3 surface cell (from CIF, after surface(0,0,1))
a_cr = 5.004  # Å
b_cr = 4.333  # Å

# Hastelloy N FCC cubic
a_ni = 3.524  # Å

print("=" * 65)
print(f"Cr2O3  surface cell: {a_cr:.3f} x {b_cr:.3f} Å")
print(f"Hastelloy N unit cell: {a_ni:.3f} x {a_ni:.3f} Å")
print("=" * 65)
print(f"\n{'Cr repeat':>10} {'HN repeat':>10} {'Cr x':>8} {'Cr y':>8} "
      f"{'HN x':>8} {'HN y':>8} {'strain x':>10} {'strain y':>10} {'total':>8}")
print("-" * 95)

results = []

for i in range(1, 6):      # Cr2O3 x repeat
    for j in range(1, 6):  # Cr2O3 y repeat
        for k in range(1, 7):  # HN x repeat
            for l in range(1, 7):  # HN y repeat

                cr_x = i * a_cr
                cr_y = j * b_cr
                hn_x = k * a_ni
                hn_y = l * a_ni

                # Strain = (HN - Cr) / Cr * 100
                sx = abs(hn_x - cr_x) / cr_x * 100
                sy = abs(hn_y - cr_y) / cr_y * 100
                total = sx + sy

                # Only keep if total strain < 10% and size ~10-20 Å
                if (total < 10.0 and
                    10.0 < cr_x < 22.0 and
                    10.0 < cr_y < 22.0):
                    results.append((total, sx, sy,
                                    i, j, k, l,
                                    cr_x, cr_y, hn_x, hn_y))

# Sort by total strain
results.sort(key=lambda x: x[0])

# Print top 10
for r in results[:10]:
    total, sx, sy, i, j, k, l, cr_x, cr_y, hn_x, hn_y = r
    print(f"  Cr {i}x{j}    HN {k}x{l}    "
          f"{cr_x:7.2f}  {cr_y:7.2f}  "
          f"{hn_x:7.2f}  {hn_y:7.2f}  "
          f"{sx:9.1f}%  {sy:9.1f}%  {total:7.1f}%")

print("\n" + "=" * 65)
print("Top recommendation: lowest total strain pair above")
print("=" * 65)
```

```python
from ase.spacegroup import crystal
from ase.io import write
from ase.neighborlist import natural_cutoffs, NeighborList
from ase.build import niggli_reduce
import numpy as np

# ── Build Cr2O3 bulk unit cell ────────────────────────────
a = 4.96
c = 13.59

bulk_cr2o3 = crystal(
    symbols=['Cr', 'O'],
    basis=[(0, 0, 0.347), (0.306, 0, 0.25)],
    spacegroup=167,
    cellpar=[a, a, c, 90, 90, 120]
)

print(f"Unit cell: {bulk_cr2o3.get_chemical_formula()}, {len(bulk_cr2o3)} atoms")
print(f"Unit cell matrix:\n{bulk_cr2o3.get_cell()}")

# ── Repeat to get coincidence supercell size ──────────────
# Target: ~10 x 17 Å lateral to match Hastelloy N 3x5 cell
slab = bulk_cr2o3.repeat([2, 4, 1])

print(f"\nAfter 2×4×1 repeat:")
print(f"  Atoms:  {len(slab)}")
print(f"  Cell matrix:\n{slab.get_cell()}")

# ── Apply Niggli reduction ────────────────────────────────
niggli_reduce(slab)
print(f"\nAfter Niggli reduction:")
print(f"  Cell matrix:\n{slab.get_cell()}")

# ── Convert to orthogonal ─────────────────────────────────
cell = slab.get_cell()
Lx = cell[0, 0]
Ly = cell[1, 1]
Lz = cell[2, 2]
slab.set_cell(np.diag([Lx, Ly, Lz]), scale_atoms=True)
slab.set_pbc([True, True, True])
slab.wrap()

# ── Verify ────────────────────────────────────────────────
cell = slab.get_cell()
z_pos = slab.get_positions()[:, 2]
thickness = z_pos.max() - z_pos.min()

print(f"\n=== Cr2O3 Bulk Slab (Adjusted) ===")
print(f"Atoms:          {len(slab)}")
print(f"Formula:        {slab.get_chemical_formula()}")
print(f"Cell (Å):       {cell[0,0]:.3f} x {cell[1,1]:.3f} x {cell[2,2]:.3f}")
print(f"Thickness:      {thickness:.2f} Å")
print(f"PBC:            {slab.get_pbc()}")
print(f"Cell matrix (should be diagonal):\n{slab.get_cell()}")

# ── Verify Cr-O bond lengths ──────────────────────────────
cutoffs = natural_cutoffs(slab, mult=1.2)
nl = NeighborList(cutoffs, self_interaction=False, bothways=True)
nl.update(slab)
symbols   = slab.get_chemical_symbols()
positions = slab.get_positions()

cr_o_bonds = []
for i, sym in enumerate(symbols):
    if sym == 'Cr':
        indices, offsets = nl.get_neighbors(i)
        for j, offset in zip(indices, offsets):
            if symbols[j] == 'O':
                d = np.linalg.norm(
                    positions[j] + offset @ slab.get_cell() - positions[i])
                cr_o_bonds.append(d)

if cr_o_bonds:
    print(f"\nCr-O bond lengths:")
    print(f"  Min: {min(cr_o_bonds):.3f} Å  (expected ~1.97 Å)")
    print(f"  Max: {max(cr_o_bonds):.3f} Å  (expected ~2.02 Å)")
else:
    print("\n⚠ No Cr-O bonds found — check structure")

# ── Write files ───────────────────────────────────────────
lammps_file = 'cr2o3_perfect_adjusted.lammps'
write(lammps_file, slab,
      format='lammps-data',
      atom_style='atomic',
      specorder=['Cr', 'O'])

with open(lammps_file, 'a') as f:
    f.write('\nMasses\n\n')
    f.write('1 51.996  # Cr\n')
    f.write('2 15.999  # O\n')

write('cr2o3_perfect_adjusted.xyz', slab)

print(f"\n✓ Files written: cr2o3_perfect_adjusted.lammps + cr2o3_perfect_adjusted.xyz")
```

```python
from ase.build import bulk
from ase.io import write
import numpy as np

# ── Step 2: FCC Ni base ──────────────────────────────────
a_Ni = 3.524  # Angstrom
base = bulk('Ni', 'fcc', a=a_Ni, cubic=True)

print(f"Unit cell:")
print(f"  Atoms:   {len(base)}")
print(f"  Cell:    {base.get_cell().diagonal()}")

# 3x5x8 repeat
slab = base.repeat([3, 5, 8])
N = len(slab)
cell = slab.get_cell()

print(f"\nAfter 3x3x8 repeat:")
print(f"  Atoms:     {N}")
print(f"  Cell (Å):  {cell[0,0]:.2f} x {cell[1,1]:.2f} x {cell[2,2]:.2f}")
print(f"  Thickness: {cell[2,2]:.2f} Å")

# ── Step 3: Composition wt% → at% ────────────────────────
n_Mo = int(0.107 * N)
n_Cr = int(0.088 * N)
n_Fe = int(0.041 * N)
n_C  = int(0.0038 * N)
n_Al = int(0.0106 * N)
n_B  = int(0.0005 * N)

total_solute = n_Mo + n_Cr + n_Fe + n_C + n_Al + n_B
n_Ni = N - total_solute

print(f"\nComposition:")
print(f"  Ni: {n_Ni}  ({100*n_Ni/N:.1f} at%)")
print(f"  Mo: {n_Mo}  ({100*n_Mo/N:.1f} at%)")
print(f"  Cr: {n_Cr}  ({100*n_Cr/N:.1f} at%)")
print(f"  Fe: {n_Fe}  ({100*n_Fe/N:.1f} at%)")
print(f"  C:  {n_C}   ({100*n_C/N:.2f} at%)")
print(f"  Al: {n_Al}  ({100*n_Al/N:.2f} at%)")
print(f"  B:  {n_B}   ({100*n_B/N:.3f} at%)")
print(f"  Total solute: {total_solute}")

# ── Random substitution ───────────────────────────────────
np.random.seed(42)
symbols = np.array(slab.get_chemical_symbols())
indices = np.random.permutation(N)
print(symbols)
print(indices)

idx = 0
for element, count in [('Mo', n_Mo), ('Cr', n_Cr),
                        ('Fe', n_Fe), ('C',  n_C),
                        ('Al', n_Al), ('B',  n_B)]:
    symbols[indices[idx:idx+count]] = element
    idx += count

slab.set_chemical_symbols(symbols.tolist())

# ── Step 4: Add vacuum ────────────────────────────────────
slab.center(vacuum=5.0, axis=2)

# ── Step 5: Verify ────────────────────────────────────────
cell = slab.get_cell()
z_pos = slab.get_positions()[:, 2]
thickness = z_pos.max() - z_pos.min()
vacuum_actual = cell[2,2] - thickness

print(f"\nFinal slab:")
print(f"  Cell (Å):  {cell[0,0]:.2f} x {cell[1,1]:.2f} x {cell[2,2]:.2f}")
print(f"  Thickness: {thickness:.1f} Å")
print(f"  Vacuum:    {vacuum_actual:.1f} Å")
print(f"  Formula:   {slab.get_chemical_formula()}")

# ── Step 6: Write output files ────────────────────────────
lammps_file = 'hastelloy_n_perfect_adjusted.lammps'
specorder = ['Ni', 'Mo', 'Cr', 'Fe', 'C', 'Al', 'B']
write(lammps_file, slab,
      format='lammps-data',
      atom_style='atomic',
      specorder=specorder)

# Append Masses
with open(lammps_file, 'a') as f:
    f.write('\nMasses\n\n')
    f.write('1  58.693  # Ni\n')
    f.write('2  95.960  # Mo\n')
    f.write('3  51.996  # Cr\n')
    f.write('4  55.845  # Fe\n')
    f.write('5  12.011  # C\n')
    f.write('6  26.982  # Al\n')
    f.write('7  10.811  # B\n')

write('hastelloy_n_perfect_adjusted.xyz', slab)

print(f"\n=== Files Written ===")
print(f"  hastelloy_n_perfect_adjusted.lammps")
print(f"  hastelloy_n_perfect_adjusted.xyz")
print(f"\n✓ Ready for OVITO visualization")
```

```python
from ase.spacegroup import crystal
from ase.build import niggli_reduce
from ase.io import read, write
from ase import Atoms
import numpy as np

# ── Build Cr2O3 ───────────────────────────────────────────
a = 4.96
c = 13.59

bulk_cr2o3 = crystal(
    symbols=['Cr', 'O'],
    basis=[(0, 0, 0.347), (0.306, 0, 0.25)],
    spacegroup=167,
    cellpar=[a, a, c, 90, 90, 120]
)

cr2o3 = bulk_cr2o3.repeat([2, 4, 2])
niggli_reduce(cr2o3)
cr2o3.set_pbc([True, True, True])
cr2o3.wrap()

# Shift Cr2O3 atoms to start at z=0
z_cr = cr2o3.get_positions()[:, 2]
pos_cr = cr2o3.get_positions().copy()
pos_cr[:, 2] -= z_cr.min()
cr2o3.set_positions(pos_cr)
cr2o3_thickness = pos_cr[:, 2].max()

print(f"Cr2O3:  {len(cr2o3)} atoms")
print(f"  z: 0.000 – {cr2o3_thickness:.3f} Å")

# ── Load Hastelloy N ──────────────────────────────────────
hastn = read('Molecular_dynamics/structures/HastellotN_perfect/hastelloy_n_perfect_adjusted.xyz')
hastn.set_pbc([True, True, True])

# Shift Hastelloy N atoms to start at z=0
z_hn = hastn.get_positions()[:, 2]
pos_hn = hastn.get_positions().copy()
pos_hn[:, 2] -= z_hn.min()
hastn.set_positions(pos_hn)
hastn_thickness = pos_hn[:, 2].max()

print(f"\nHastelloy N: {len(hastn)} atoms")
print(f"  z: 0.000 – {hastn_thickness:.3f} Å")

# ── Strain Cr2O3 x-y to match Hastelloy N ─────────────────
hn_Lx = hastn.get_cell()[0, 0]
hn_Ly = hastn.get_cell()[1, 1]
cr_Lx = cr2o3.get_cell()[0, 0]
cr_Ly = cr2o3.get_cell()[1, 1]

sx = hn_Lx / cr_Lx
sy = hn_Ly / cr_Ly
print(f"\nStrain applied to Cr2O3:")
print(f"  x: {(sx-1)*100:.1f}%")
print(f"  y: {(sy-1)*100:.1f}%")

# Apply strain to Cr2O3 x-y positions only
pos_cr = cr2o3.get_positions().copy()
pos_cr[:, 0] *= sx
pos_cr[:, 1] *= sy

# ── Shift Cr2O3 above Hastelloy N with gap ────────────────
gap = 2.0
pos_cr[:, 2] += hastn_thickness + gap

print(f"\nAfter shift — Cr2O3 z: {pos_cr[:,2].min():.3f} – {pos_cr[:,2].max():.3f} Å")
print(f"Expected bottom of Cr2O3: {hastn_thickness + gap:.3f} Å")

# ── Calculate total cell height FIRST ────────────────────
total_z = pos_cr[:, 2].max() + 2.0

# ── Combine structures ────────────────────────────────────
n_hastn = len(hastn)
n_cr2o3 = len(cr2o3)

all_symbols = hastn.get_chemical_symbols() + cr2o3.get_chemical_symbols()
all_positions = np.vstack([pos_hn, pos_cr])

L1 = Atoms(symbols=all_symbols,
           positions=all_positions,
           cell=[hn_Lx, hn_Ly, total_z],
           pbc=[True, True, True])

# ── Verify ────────────────────────────────────────────────
# Metal indices are the first n_hastn atoms
# Oxide indices are the last n_cr2o3 atoms
z_alloy = all_positions[:n_hastn, 2]
z_oxide = all_positions[n_hastn:, 2]

print(f"\n=== L1 Combined Structure ===")
print(f"Total atoms:     {len(L1)}")
cell = L1.get_cell()
print(f"Cell (Å):        {cell[0,0]:.3f} x {cell[1,1]:.3f} x {cell[2,2]:.3f}")
print(f"\nHastelloy N z:   {z_alloy.min():.3f} – {z_alloy.max():.3f} Å")
print(f"Cr2O3 z:         {z_oxide.min():.3f} – {z_oxide.max():.3f} Å")
print(f"Interface gap:   {z_oxide.min() - z_alloy.max():.3f} Å  (target 2.0 Å)")

# ── Write files ───────────────────────────────────────────
specorder = ['Ni', 'Mo', 'Cr', 'Fe', 'C', 'Al', 'B', 'O']
lammps_file = 'L1_perfect.lammps'
write(lammps_file, L1,
      format='lammps-data',
      atom_style='atomic',
      specorder=specorder)

with open(lammps_file, 'a') as f:
    f.write('\nMasses\n\n')
    f.write('1  58.693  # Ni\n')
    f.write('2  95.960  # Mo\n')
    f.write('3  51.996  # Cr\n')
    f.write('4  55.845  # Fe\n')
    f.write('5  12.011  # C\n')
    f.write('6  26.982  # Al\n')
    f.write('7  10.811  # B\n')
    f.write('8  15.999  # O\n')

write('L1_perfect.xyz', L1)

print(f"\n=== Files Written ===")
print(f"  L1_perfect.lammps")
print(f"  L1_perfect.xyz")
print(f"\n✓ Ready for OVITO visualization")
```

```python
from ase.build import bulk
from ase.io import write
from ase.neighborlist import natural_cutoffs, NeighborList
import numpy as np

a_NiO = 4.177

# ── cubic=True gives orthogonal cell ─────────────────────
NiO_bulk = bulk('NiO', 'rocksalt', a=a_NiO, cubic=True)

print(f"NiO unit cell:")
print(f"  Atoms:   {len(NiO_bulk)}")
print(f"  Formula: {NiO_bulk.get_chemical_formula()}")
print(f"  Cell (Å): {NiO_bulk.get_cell().diagonal()}")
print(f"  Cell matrix:\n{NiO_bulk.get_cell()}")

# ── Test repeat counts ────────────────────────────────────
print(f"\nRepeat options:")
for nx, ny, nz in [(2,3,3),(2,4,3),(3,4,3),(2,4,2)]:
    slab = NiO_bulk.repeat([nx, ny, nz])
    cell = slab.get_cell()
    z_pos = slab.get_positions()[:, 2]
    thick = z_pos.max() - z_pos.min()
    print(f"  {nx}x{ny}x{nz}: {len(slab):>4} atoms, "
          f"{cell[0,0]:.2f} x {cell[1,1]:.2f} x {cell[2,2]:.2f} Å, "
          f"thickness={thick:.2f} Å")

# ── Build final slab ──────────────────────────────────────
# Target: ~10-12 Å lateral (match coincidence cell), ~13 Å thick
slab = NiO_bulk.repeat([2, 4, 3])
slab.set_pbc([True, True, True])

cell  = slab.get_cell()
z_pos = slab.get_positions()[:, 2]
thickness = z_pos.max() - z_pos.min()

print(f"\n=== NiO Slab ===")
print(f"Atoms:     {len(slab)}")
print(f"Formula:   {slab.get_chemical_formula()}")
print(f"Cell (Å):  {cell[0,0]:.3f} x {cell[1,1]:.3f} x {cell[2,2]:.3f}")
print(f"Thickness: {thickness:.3f} Å")
print(f"PBC:       {slab.get_pbc()}")
print(f"Cell matrix:\n{slab.get_cell()}")

# ── Verify Ni-O bond lengths ──────────────────────────────
cutoffs   = natural_cutoffs(slab, mult=1.2)
nl        = NeighborList(cutoffs, self_interaction=False, bothways=True)
nl.update(slab)
symbols   = slab.get_chemical_symbols()
positions = slab.get_positions()

ni_o_bonds = []
for i, sym in enumerate(symbols):
    if sym == 'Ni':
        indices, offsets = nl.get_neighbors(i)
        for j, offset in zip(indices, offsets):
            if symbols[j] == 'O':
                d = np.linalg.norm(
                    positions[j] + offset @ slab.get_cell() - positions[i])
                ni_o_bonds.append(d)

print(f"\nNi-O bond lengths:")
print(f"  Min: {min(ni_o_bonds):.3f} Å  (expected ~2.09 Å)")
print(f"  Max: {max(ni_o_bonds):.3f} Å  (expected ~2.09 Å)")

# ── Write files ───────────────────────────────────────────
lammps_file = 'NiO_perfect.lammps'
write(lammps_file, slab,
      format='lammps-data',
      atom_style='atomic',
      specorder=['Ni', 'O'])

with open(lammps_file, 'a') as f:
    f.write('\nMasses\n\n')
    f.write('1  58.693  # Ni\n')
    f.write('2  15.999  # O\n')

write('NiO_perfect.xyz', slab)

print(f"\n=== Files Written ===")
print(f"  NiO_perfect.lammps")
print(f"  NiO_perfect.xyz")
print(f"\n✓ Ready for OVITO visualization")
```

```python
from ase.spacegroup import crystal
from ase.build import bulk, niggli_reduce
from ase.io import write
import numpy as np

# ── Build Cr2O3 with correct axis orientation ─────────────
a_cr = 4.96
c_cr = 13.59

bulk_cr2o3 = crystal(
    symbols=['Cr', 'O'],
    basis=[(0, 0, 0.347), (0.306, 0, 0.25)],
    spacegroup=167,
    cellpar=[a_cr, a_cr, c_cr, 90, 90, 120]
)

# Use repeat [4, 2, 1] instead of [2, 4, 1]
# so after niggli_reduce the axes come out correctly
# x=17.18, y=9.92, z=13.59 → then we rotate z to be coating direction
cr2o3 = bulk_cr2o3.repeat([2, 4, 1])
niggli_reduce(cr2o3)
cr2o3.set_pbc([True, True, True])
cr2o3.wrap()

# ── Swap axes: make z = coating thickness direction ───────
# Current: x=9.92, y=13.59, z=17.18
# Target:  x=9.92, y=17.18, z=13.59 (swap y and z)
from ase import Atoms

pos  = cr2o3.get_positions().copy()
cell = cr2o3.get_cell().copy()

# Swap y and z columns in positions
pos_new       = pos.copy()
pos_new[:, 1] = pos[:, 2]   # new y = old z
pos_new[:, 2] = pos[:, 1]   # new z = old y

# Swap y and z in cell
cell_new      = np.diag([cell[0,0], cell[2,2], cell[1,1]])

cr2o3_fixed = Atoms(
    symbols   = cr2o3.get_chemical_symbols(),
    positions = pos_new,
    cell      = cell_new,
    pbc       = [True, True, True]
)
cr2o3_fixed.wrap()

# Shift to z=0
z_cr = cr2o3_fixed.get_positions()[:, 2]
pos  = cr2o3_fixed.get_positions().copy()
pos[:, 2] -= z_cr.min()
cr2o3_fixed.set_positions(pos)
cr2o3_thickness = pos[:, 2].max()

cell_cr = cr2o3_fixed.get_cell()
print(f"Cr2O3 (axis-corrected):")
print(f"  Atoms:     {len(cr2o3_fixed)}")
print(f"  Cell (Å):  {cell_cr[0,0]:.3f} x {cell_cr[1,1]:.3f} x {cell_cr[2,2]:.3f}")
print(f"  Thickness: {cr2o3_thickness:.3f} Å")

# ── Build NiO ─────────────────────────────────────────────
a_NiO    = 4.177
NiO_bulk = bulk('NiO', 'rocksalt', a=a_NiO, cubic=True)
nio      = NiO_bulk.repeat([2, 4, 3])
nio.set_pbc([True, True, True])

z_nio = nio.get_positions()[:, 2]
pos   = nio.get_positions().copy()
pos[:, 2] -= z_nio.min()
nio.set_positions(pos)
nio_thickness = pos[:, 2].max()

cell_nio = nio.get_cell()
print(f"\nNiO:")
print(f"  Atoms:     {len(nio)}")
print(f"  Cell (Å):  {cell_nio[0,0]:.3f} x {cell_nio[1,1]:.3f} x {cell_nio[2,2]:.3f}")
print(f"  Thickness: {nio_thickness:.3f} Å")

# ── Reference lateral cell from Hastelloy N ───────────────
ref_Lx = 10.572
ref_Ly = 17.620

# ── Strain Cr2O3 x-y ──────────────────────────────────────
sx_cr = ref_Lx / cell_cr[0, 0]
sy_cr = ref_Ly / cell_cr[1, 1]
pos_cr = cr2o3_fixed.get_positions().copy()
pos_cr[:, 0] *= sx_cr
pos_cr[:, 1] *= sy_cr
cr2o3_fixed.set_positions(pos_cr)
print(f"\nCr2O3 strain: x={(sx_cr-1)*100:.1f}%, y={(sy_cr-1)*100:.1f}%")

# ── Strain NiO x-y ────────────────────────────────────────
sx_nio = ref_Lx / cell_nio[0, 0]
sy_nio = ref_Ly / cell_nio[1, 1]
pos_nio = nio.get_positions().copy()
pos_nio[:, 0] *= sx_nio
pos_nio[:, 1] *= sy_nio

# ── Place NiO above Cr2O3 with 2 Å gap ───────────────────
gap = 2.0
pos_nio[:, 2] += cr2o3_thickness + gap
nio.set_positions(pos_nio)
print(f"NiO strain:   x={(sx_nio-1)*100:.1f}%, y={(sy_nio-1)*100:.1f}%")

# ── Combine ───────────────────────────────────────────────
O2 = cr2o3_fixed + nio
total_z = nio.get_positions()[:, 2].max() + 2.0
O2.set_cell([ref_Lx, ref_Ly, total_z])
O2.set_pbc([True, True, True])

# ── Verify ────────────────────────────────────────────────
positions = O2.get_positions()
symbols   = O2.get_chemical_symbols()

cr_idx = [i for i,s in enumerate(symbols) if s == 'Cr']
ni_idx = [i for i,s in enumerate(symbols) if s == 'Ni']

z_cr_atoms = positions[cr_idx, 2]
z_ni_atoms = positions[ni_idx, 2]

print(f"\n=== O2: NiO + Cr2O3 Bilayer ===")
print(f"Total atoms:             {len(O2)}")
cell = O2.get_cell()
print(f"Cell (Å):                {cell[0,0]:.3f} x {cell[1,1]:.3f} x {cell[2,2]:.3f}")
print(f"\nCr2O3 layer z:           {z_cr_atoms.min():.2f} – {z_cr_atoms.max():.2f} Å")
print(f"NiO layer z:             {z_ni_atoms.min():.2f} – {z_ni_atoms.max():.2f} Å")
print(f"Interface gap:           {z_ni_atoms.min() - z_cr_atoms.max():.2f} Å (target 2.0 Å)")
print(f"Total bilayer thickness: {z_ni_atoms.max() - z_cr_atoms.min():.2f} Å")

# ── Write files ───────────────────────────────────────────
specorder   = ['Cr', 'Ni', 'O']
lammps_file = 'O2_bilayer.lammps'
write(lammps_file, O2,
      format='lammps-data',
      atom_style='atomic',
      specorder=specorder)

with open(lammps_file, 'a') as f:
    f.write('\nMasses\n\n')
    f.write('1  51.996  # Cr\n')
    f.write('2  58.693  # Ni\n')
    f.write('3  15.999  # O\n')

write('O2_bilayer.xyz', O2)

print(f"\n=== Files Written ===")
print(f"  O2_bilayer.lammps")
print(f"  O2_bilayer.xyz")
print(f"\n✓ Ready for OVITO visualization")
```
